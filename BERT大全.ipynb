```python
import pandas as pd
from transformers import BertTokenizer, BertForMaskedLM
import torch
from tqdm import tqdm
import numpy as np

# 加载预训练的BERT模型和分词器
tokenizer = BertTokenizer.from_pretrained('bert-base-chinese')
model = BertForMaskedLM.from_pretrained('bert-base-chinese')

# 读取数据集
df = pd.read_csv('selected_dev_dataset.csv')

results = []

# 使用tqdm创建进度条
for index, row in tqdm(df.iterrows(), total=len(df), desc="Processing rows"):
    sentence = row['tagged']  # 使用'tagged'列作为句子
    correct_word = row['classifier']  # 使用'classifier'列作为正确答案

    # 替换<CL>为[MASK]，替换<h>和</h>为空字符串
    sentence = sentence.replace('<CL>', '[MASK]').replace('<h>', '').replace('</h>', '')

    # 分词
    inputs = tokenizer(sentence, return_tensors='pt')

    # 获取预测结果
    with torch.no_grad():
        outputs = model(**inputs)
        logits = outputs.logits  # 原始logits（未归一化的分数）

    # 获取[MASK]位置的预测结果
    mask_token_index = torch.where(inputs['input_ids'] == tokenizer.mask_token_id)[1]

    if mask_token_index.numel() == 0:
        print(f"\nWarning: No [MASK] token found in sentence: {sentence}")
        # 如果没有找到[MASK]，记录特殊的值
        top1_word = "N/A"
        top1_logprob = "N/A"
        top3_words = ["N/A", "N/A", "N/A"]
        top3_logprobs = ["N/A", "N/A", "N/A"]
    else:
        # 获取[MASK]位置的logits
        mask_logits = logits[0, mask_token_index[0]]

        # 计算softmax概率
        probs = torch.nn.functional.softmax(mask_logits, dim=-1)

        # 计算对数概率（log probabilities）
        log_probs = torch.log(probs)

        # 获取topk预测结果 (k=3)
        topk_logprobs, topk_token_ids = torch.topk(log_probs, k=3)
        topk_logprobs = topk_logprobs.tolist()
        topk_token_ids = topk_token_ids.tolist()

        # 解码token
        topk_words = tokenizer.convert_ids_to_tokens(topk_token_ids)

        # 获取top1结果
        top1_word = topk_words[0]
        top1_logprob = topk_logprobs[0]

        # 获取top3结果
        top3_words = topk_words[:3]
        top3_logprobs = topk_logprobs[:3]

    # 将结果保存到列表中
    results.append([
        row['new_index'],
        sentence,
        correct_word,
        top1_word,
        top1_logprob,
        ", ".join(top3_words),
        ", ".join([f"{lp:.4f}" for lp in top3_logprobs])
    ])

    # 打印当前处理行的信息
    print(f"\nProcessed row {index + 1}/{len(df)}")
    print(f"Sentence: {sentence}")
    print(f"Target: {correct_word}")
    print(f"Top1 prediction: {top1_word} (logprob: {top1_logprob:.4f})")
    print(f"Top3 predictions: {', '.join(top3_words)}")
    print(f"Top3 log probabilities: {', '.join([f'{lp:.4f}' for lp in top3_logprobs])}")

# 将结果保存到新的CSV文件
result_df = pd.DataFrame(results, columns=[
    'new_index',
    'sentence',
    'target',
    'top1_prediction',
    'top1_log_probability',
    'top3_predictions',
    'top3_log_probabilities'
])
result_df.to_csv('bert_mlm_result.csv', index=False, encoding='utf-8-sig')

print("\nResults saved to bert_mlm_result.csv")
```

```python
import pandas as pd
from transformers import BertTokenizer, BertForMaskedLM
import torch
from tqdm import tqdm
import numpy as np

# 加载预训练的BERT模型和分词器
tokenizer = BertTokenizer.from_pretrained('bert-base-chinese')
model = BertForMaskedLM.from_pretrained('bert-base-chinese')

# 读取数据集
df = pd.read_csv('selected_dev_dataset.csv')

results = []

# 使用tqdm创建进度条
for index, row in tqdm(df.iterrows(), total=len(df), desc="Processing rows"):
    sentence = row['tagged']  # 使用'tagged'列作为句子
    correct_word = row['classifier']  # 使用'classifier'列作为正确答案

    # 提取<h>和</h>之间的头词，并移除<h>和</h>以及头词
    head_word_start = sentence.find('<h>')
    head_word_end = sentence.find('</h>')
    if head_word_start != -1 and head_word_end != -1:
        head_word = sentence[head_word_start + 3:head_word_end]  # 提取<h>和</h>之间的词
        sentence = sentence[:head_word_start] + sentence[head_word_end + 4:]  # 移除<h>、</h>和头词
    else:
        head_word = None  # 如果没有找到<h>和</h>，则头词为空

    # 替换<CL>为[MASK]
    sentence = sentence.replace('<CL>', '[MASK]')

    # 分词
    inputs = tokenizer(sentence, return_tensors='pt')

    # 获取头词的位置
    if head_word is not None:
        head_word_index = torch.where(inputs['input_ids'] == tokenizer.convert_tokens_to_ids(head_word))[1]
    else:
        head_word_index = torch.tensor([])  # 如果没有头词，则创建一个空的张量

    # 创建注意力掩码
    attention_mask = torch.ones_like(inputs['input_ids'])
    if head_word_index.numel() > 0:
        attention_mask[:, head_word_index] = 0  # 遮蔽头词的注意力

    # 更新inputs中的attention_mask
    inputs['attention_mask'] = attention_mask

    # 获取预测结果
    with torch.no_grad():
        outputs = model(**inputs)
        logits = outputs.logits  # 原始logits（未归一化的分数）

    # 获取[MASK]位置的预测结果
    mask_token_index = torch.where(inputs['input_ids'] == tokenizer.mask_token_id)[1]

    if mask_token_index.numel() == 0:
        print(f"\nWarning: No [MASK] token found in sentence: {sentence}")
        top1_word = "N/A"
        top1_logprob = "N/A"
        top3_words = ["N/A", "N/A", "N/A"]
        top3_logprobs = ["N/A", "N/A", "N/A"]
    else:
        mask_logits = logits[0, mask_token_index[0]]
        probs = torch.nn.functional.softmax(mask_logits, dim=-1)
        log_probs = torch.log(probs)
        topk_logprobs, topk_token_ids = torch.topk(log_probs, k=3)
        topk_logprobs = topk_logprobs.tolist()
        topk_token_ids = topk_token_ids.tolist()
        topk_words = tokenizer.convert_ids_to_tokens(topk_token_ids)
        top1_word = topk_words[0]
        top1_logprob = topk_logprobs[0]
        top3_words = topk_words[:3]
        top3_logprobs = topk_logprobs[:3]

    results.append([
        row['new_index'],
        sentence,
        correct_word,
        top1_word,
        top1_logprob,
        ", ".join(top3_words),
        ", ".join([f"{lp:.4f}" for lp in top3_logprobs])
    ])

result_df = pd.DataFrame(results, columns=[
    'new_index',
    'sentence',
    'target',
    'top1_prediction',
    'top1_log_probability',
    'top3_predictions',
    'top3_log_probabilities'
])
result_df.to_csv('bert_mlm_head_masking_result.csv', index=False, encoding='utf-8-sig')
print("\nResults saved to bert_mlm_head_masking_result.csv")
```

```python
import pandas as pd
from transformers import BertTokenizer, BertForMaskedLM
import torch
from tqdm import tqdm

# 加载预训练的BERT模型和分词器
tokenizer = BertTokenizer.from_pretrained('bert-base-chinese')
model = BertForMaskedLM.from_pretrained('bert-base-chinese')

# 定义允许的172个量词集合
allowed_classifiers = {
    '个', '种', '次', '张', '件', '句', '条', '位', '家', '场', '只', '点', '部', '首',
    '篇', '段', '号', '份', '块', '滴', '颗', '群', '名', '款', '片', '级', '堆', '些',
    '本', '分', '杯', '起', '步', '把', '顿', '对', '碗', '套', '集', '辆', '回', '代',
    '座', '口', '道', '层', '声', '类', '届', '班', '项', '番', '阵', '支', '双', '等',
    '股', '瓶', '台', '丝', '封', '根', '朵', '趟', '副', '轮', '遍', '头', '包', '幅',
    '粒', '组', '门', '枚', '批', '盘', '棵', '间', '样', '季', '笔', '波', '面', '楼',
    '盆', '节', '身', '桌', '盒', '下', '页', '箱', '系列', '排', '处', '袋', '锅', '通',
    '所', '栋', '团', '盏', '发', '串', '圈', '束', '桶', '枝', '餐', '世', '堂', '则',
    '壶', '架', '手', '码', '艘', '匹', '幕', '扇', '户', '线', '曲', '伙', '刀', '环',
    '肚子', '卷', '株', '笼', '任', '罐', '脸', '行', '般', '顶', '桩', '袭', '版', '路',
    '章', '记', '拨', '辈', '杆', '尾', '味', '剂', '列', '具', '册', '幢', '例', '队',
    '帖', '尊', '缸', '拳', '局', '宗', '遭', '席', '人次', '档子', '管', '员', '棒',
    '缕', '帮', '族', '堵', '眼'
}

# 获取这些量词对应的token ID（只考虑单token的量词）
allowed_token_ids = set()
for word in allowed_classifiers:
    tokens = tokenizer.tokenize(word)
    if len(tokens) == 1:  # 只考虑单token的量词
        allowed_token_ids.add(tokenizer.convert_tokens_to_ids(tokens)[0])

# 读取数据集
df = pd.read_csv('selected_dev_dataset.csv')
results = []

# 使用tqdm创建进度条
for index, row in tqdm(df.iterrows(), total=len(df), desc="Processing rows"):
    sentence = row['tagged']  # 使用'tagged'列作为句子
    correct_word = row['classifier']  # 使用'classifier'列作为正确答案

    # 提取<h>和</h>之间的头词，并移除<h>和</h>以及头词
    head_word_start = sentence.find('<h>')
    head_word_end = sentence.find('</h>')
    if head_word_start != -1 and head_word_end != -1:
        head_word = sentence[head_word_start + 3:head_word_end]  # 提取<h>和</h>之间的词
        sentence = sentence[:head_word_start] + sentence[head_word_end + 4:]  # 移除<h>、</h>和头词
    else:
        head_word = None  # 如果没有找到<h>和</h>，则头词为空

    # 替换<CL>为[MASK]
    sentence = sentence.replace('<CL>', '[MASK]')

    # 分词
    inputs = tokenizer(sentence, return_tensors='pt')

    # 获取头词的位置（如果存在）
    if head_word is not None:
        head_word_tokens = tokenizer.tokenize(head_word)
        head_word_token_ids = [tokenizer.convert_tokens_to_ids(t) for t in head_word_tokens]
        head_word_indices = []
        for token_id in head_word_token_ids:
            matches = torch.where(inputs['input_ids'] == token_id)[1]
            head_word_indices.extend(matches.tolist())
    else:
        head_word_indices = []

    # 创建注意力掩码（默认全部关注）
    attention_mask = torch.ones_like(inputs['input_ids'])
    # 遮蔽头词的注意力（如果存在）
    if head_word_indices:
        attention_mask[:, head_word_indices] = 0

    # 更新inputs中的attention_mask
    inputs['attention_mask'] = attention_mask

    # 获取预测结果
    with torch.no_grad():
        outputs = model(**inputs)
        logits = outputs.logits  # 原始logits（未归一化的分数）

    # 获取[MASK]位置的预测结果
    mask_token_index = torch.where(inputs['input_ids'] == tokenizer.mask_token_id)[1]

    if mask_token_index.numel() == 0:
        print(f"\nWarning: No [MASK] token found in sentence: {sentence}")
        top1_word = "N/A"
        top1_logprob = "N/A"
        top3_words = ["N/A", "N/A", "N/A"]
        top3_logprobs = ["N/A", "N/A", "N/A"]
    else:
        # 获取[MASK]位置的logits
        mask_logits = logits[0, mask_token_index[0]]

        # 创建限制掩码，只允许172个量词的token
        restrict_mask = torch.zeros_like(mask_logits)
        for token_id in allowed_token_ids:
            restrict_mask[token_id] = 1

        # 应用限制，将不允许的token的logits设为负无穷
        restricted_logits = mask_logits * restrict_mask + (1 - restrict_mask) * -1e10

        # 计算对数概率
        probs = torch.nn.functional.softmax(restricted_logits, dim=-1)
        log_probs = torch.log(probs)

        # 获取topk预测结果 (k=3)
        topk_logprobs, topk_token_ids = torch.topk(log_probs, k=3)
        topk_words = tokenizer.convert_ids_to_tokens(topk_token_ids.tolist())

        top1_word = topk_words[0]
        top1_logprob = topk_logprobs[0].item()
        top3_words = topk_words[:3]
        top3_logprobs = [lp.item() for lp in topk_logprobs[:3]]

    results.append([
        row['new_index'],
        sentence,
        correct_word,
        top1_word,
        top1_logprob,
        ", ".join(top3_words),
        ", ".join([f"{lp:.4f}" for lp in top3_logprobs])
    ])

# 保存结果
result_df = pd.DataFrame(results, columns=[
    'new_index',
    'sentence',
    'target',
    'top1_prediction',
    'top1_log_probability',
    'top3_predictions',
    'top3_log_probabilities'
])
result_df.to_csv('bert_mlm_restricted_head_masking_result.csv', index=False, encoding='utf-8-sig')

print("\nResults saved to bert_mlm_restricted_head_masking_result.csv")

# 计算准确率
top1_acc = (result_df['top1_prediction'] == result_df['target']).mean()
top3_acc = result_df.apply(
    lambda x: x['target'] in x['top3_predictions'].split(", "),
    axis=1
).mean()

print(f"\nFinal Accuracy (Restricted to 172 Classifiers with Head Masking):")
print(f"Top-1: {top1_acc:.4f}")
print(f"Top-3: {top3_acc:.4f}")
```
